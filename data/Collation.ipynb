```python
# import libraries
import csv 
import pandas as pd 
```

Open all datasets

```python
df1 = pd.read_csv('./LSTM_data/super_sms_dataset.csv')
df1.columns = ['text', 'label']
df1 = df1.dropna()
df1['label'] = df1['label'].astype(int)
# df1['url'] = df1['text'].str.contains(r'(https?://|www\.)', regex=True, na=False).astype(int)

df1['url'] = df1['text'].str.extract(
    r'((?:https?://|www\.)[^\s]+)',
    expand=False
).fillna("")
df1
print(df1[(df1['url'] == "") & (df1['label'] == 1)])

```

```python
df2 = pd.read_csv('./LSTM_data/Mendeley_Dataset_5971.csv')
df2['label'] = df2['LABEL'].apply(lambda x: 1 if x in ['smishing', 'spam'] else 0)
df2 = df2.drop(columns=['EMAIL', 'PHONE', 'LABEL'])
df2.columns = ['text', 'url', 'label']
# df2['url'] = df2['url'].apply(lambda x: 1 if x == 'yes' else 0)
df2['url'] = df2['text'].str.extract(
    r'((?:https?://|www\.)[^\s]+)',
    expand=False
).fillna("")
df2
df2[df2['url'] == ""]
```

```python
df3 = pd.read_csv('./LSTM_data/Kaggle-SMS-Spam-Collection-Dataset.csv')
df3['label'] = df3['v1'].apply(lambda x: 1 if x == 'spam' else 0)

df3 = df3.drop(columns=['v1', 'Unnamed: 2', 'Unnamed: 3', 'Unnamed: 4'])
df3.columns = ['text', 'label']
df3['url'] = df3['text'].str.contains(r'(https?://|www\.)', regex=True, na=False).astype(int)
df3['url'] = df3['text'].str.extract(
    r'((?:https?://|www\.)[^\s]+)',
    expand=False
).fillna("")
df3[df3['url'] != ""]
```

```python
df4 = pd.read_csv('./LSTM_data/github-Smishing-Dataset-IMC25_final_dataset_output-all_smishing.csv')
df4['text'] = df4['translation'].fillna(df4['text'])
df4 = df4.drop_duplicates(subset=['text'])
df4_new = df4[['text']].copy()
df4_new['label'] = 1
#df4_new['url'] = df4_new['text'].str.contains(r'(https?://|www\.)', regex=True, na=False).astype(int)
df4_new['url'] = df4_new['text'].str.extract(
    r'((?:https?://|www\.)[^\s]+)',
    expand=False
).fillna(0)
df4_new[df4_new['url'] != 0]
```

```python
df5 = pd.read_csv('./LSTM_data/Mendeley_Dataset_10191_synthetic.csv')
df5['label'] = df5['LABEL'].apply(lambda x: 1 if x in ['smishing', 'spam'] else 0)
df5 = df5.drop(columns=['EMAIL', 'PHONE', 'LABEL'])
df5.columns = ['text', 'url', 'label']
df5['url'] = df5['url'].apply(lambda x: 1 if x == 'yes' else 0)

df5['url'] = df5['text'].str.extract(
    r'((?:https?://|www\.)[^\s]+)',
    expand=False
).fillna("")
df5[df5['url'] != ""]
```

```python
#collect user test samples and remove from dataset:
# extract real sms message data for user testing, then combine final dataset with synthetic 

cols = ['text', 'url','label']
dfs = [df1, df2, df3, df4_new]


dfs = [df.reindex(columns=cols) for df in dfs]

df = pd.concat(dfs, ignore_index=True)


df["text"] = df["text"].str.replace(
    r"http\S+|www\.\S+",
    "<URL>",
    regex=True)

df["text"] = df["text"].str.replace(r"\b[\w\.-]+@[\w\.-]+\.\w+\b", "<EMAIL>", regex=True)

df["text"] = df["text"].str.replace(r"\+?\d[\d\s\-]{7,}\d", "<PHONE>", regex=True)

df["label"].value_counts()

df['url'] = df['url'].replace(0, "")


# Smishing with URL
smishing_with_url = df[
    (df['label'] == 1) &
    (df['url'].str.strip() != "")
].sample(n=3, random_state=50)

print("Smishing with URL:")
print(smishing_with_url)


# Smishing without URL
smishing_without_url = df[
    (df['label'] == 1) &
    (df['url'].str.strip() == "")
].sample(n=3, random_state=50)

print("\nSmishing without URL:")
print(smishing_without_url)


# Legitimate with URL
legit_with_url = df[
    (df['label'] == 0) &
    (df['url'].str.strip() != "")
].sample(n=3, random_state=50)

print("\nLegitimate with URL:")
print(legit_with_url)


# Legitimate without URL
legit_without_url = df[
    (df['label'] == 0) &
    (df['url'].str.strip() == "")
].sample(n=3, random_state=50)

print("\nLegitimate without URL:")
print(legit_without_url)


# Combine all test samples
user_test_set = pd.concat([
    smishing_with_url,
    smishing_without_url,
    legit_with_url,
    legit_without_url
])

print("\nUser test set:")
print(user_test_set)

with pd.option_context('display.max_rows', None,
                       'display.max_columns', None,
                       'display.max_colwidth', None):

    print("=== Smishing with URL ===")
    print(smishing_with_url)

    print("\n=== Smishing without URL ===")
    print(smishing_without_url)

    print("\n=== Legitimate with URL ===")
    print(legit_with_url)

    print("\n=== Legitimate without URL ===")
    print(legit_without_url)

df = df.drop(user_test_set.index).reset_index(drop=True)

df5 = df5.reindex(columns=cols)

# Apply the same cleaning to df5
df5["text"] = df5["text"].str.replace(
    r"http\S+|www\.\S+",
    "<URL>",
    regex=True
)

df5["text"] = df5["text"].str.replace(
    r"\b[\w\.-]+@[\w\.-]+\.\w+\b",
    "<EMAIL>",
    regex=True
)

df5["text"] = df5["text"].str.replace(
    r"\+?\d[\d\s\-]{7,}\d",
    "<PHONE>",
    regex=True
)

df5['url'] = df5['url'].replace(0, "")


# Add df5 to the remaining dataset
df = pd.concat([
    df,
    df5
], ignore_index=True)

user_test_set.to_csv("user_evaluation_test_set.csv", index=False)
```

```python
#SMS messages to test explanation for LLM
# Smishing with URL
llm_smishing_with_url = df[
    (df['label'] == 1) &
    (df['url'].str.strip() != "")
].sample(n=7, random_state=50)

print("Smishing with URL:")
print(llm_smishing_with_url)


# Smishing without URL
llm_smishing_without_url = df[
    (df['label'] == 1) &
    (df['url'].str.strip() == "")
].sample(n=8, random_state=50)

print("\nSmishing without URL:")
print(llm_smishing_without_url)


# Legitimate with URL
llm_legit_with_url = df[
    (df['label'] == 0) &
    (df['url'].str.strip() != "")
].sample(n=7, random_state=50)

print("\nLegitimate with URL:")
print(llm_legit_with_url)


# Legitimate without URL
llm_legit_without_url = df[
    (df['label'] == 0) &
    (df['url'].str.strip() == "")
].sample(n=8, random_state=50)

print("\nLegitimate without URL:")
print(llm_legit_without_url)

# Combine all test samples
user_test_set = pd.concat([
    llm_smishing_with_url,
    llm_smishing_without_url,
    llm_legit_with_url,
    llm_legit_without_url
])

print("\nUser test set:")
print(user_test_set)

with pd.option_context('display.max_rows', None,
                       'display.max_columns', None,
                       'display.max_colwidth', None):

    print("=== Smishing with URL ===")
    print(llm_smishing_with_url)

    print("\n=== Smishing without URL ===")
    print(llm_smishing_without_url)

    print("\n=== Legitimate with URL ===")
    print(llm_legit_with_url)

    print("\n=== Legitimate without URL ===")
    print(llm_legit_without_url)

df = df.drop(user_test_set.index).reset_index(drop=True)

df.to_csv("combined_data.csv", index=False)

user_test_set.to_csv("llm_explanation_test_set.csv", index=False)
```
